```python
def depreciation_score(value):
    if value <= 20:
        return 10
    elif value <= 40:
        return 8
    elif value <= 60:
        return 6
    else:
        return 4

df["Depreciation_Score"] = df["depreciation(%)"].apply(depreciation_score)
```

```python
df[
[
"depreciation(%)",
"Depreciation_Score"
]
].head()
```

```python
def age_score(value):
    if value <= 3:
        return 10
    elif value <= 6:
        return 8
    elif value <= 10:
        return 6
    else:
        return 4

df["Age_Score"] = df["Vehicle_Age"].apply(age_score)
```

```python
def kms_score(value):
    if value <= 50000:
        return 10
    elif value <= 100000:
        return 8
    elif value <= 150000:
        return 6
    else:
        return 4

df["Kms_Score"] = df["Kms_Driven"].apply(kms_score)
```

```python
def ownership_score(value):
    if value == "First Owner":
        return 10
    elif value == "Second Owner":
        return 8
    elif value == "Third Owner":
        return 6
    else:
        return 4

df["Ownership_Score"] = df["Ownership"].apply(ownership_score)
```

```python
df["Value_for_Money_Score"] = (
    (df["Depreciation_Score"] * 0.40)
    +
    (df["Age_Score"] * 0.30)
    +
    (df["Kms_Score"] * 0.20)
    +
    (df["Ownership_Score"] * 0.10)
)
```

```python
df["Value_for_Money_Score"] = df["Value_for_Money_Score"].round(1)
```
